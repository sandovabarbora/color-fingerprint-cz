# Sanity check: extracted palettes vs source frames

Visual QA for the K-means LAB pipeline. For each brand, renders 5 random
frames side-by-side with their 5-color palette. Outputs land in
`outputs/figures/sanity_*.png` and are also displayed inline below.

Heavy lifting lives in `src/sanity.py` — this notebook is a thin wrapper.

```python
%matplotlib inline
import sys
from pathlib import Path

# Make `import src` work when the notebook is launched from notebooks/
sys.path.insert(0, str(Path.cwd().parent))

from IPython.display import Image, display
from src import config, sanity
```

```python
import pandas as pd

palettes = pd.read_parquet(config.PROJECT_ROOT / config.FRAMES_PARQUET.relative_to(config.PROJECT_ROOT))
brands = sorted(palettes['brand_id'].unique())
print(f'{len(brands)} brands, {palettes["frame_id"].nunique()} frames, {len(palettes)} palette rows')
brands
```

```python
# Generate (or refresh) the sanity PNGs and show them inline.
for brand_id in brands:
    out = sanity.render_brand_sanity(brand_id, palettes, n_frames=5, seed=0)
    display(Image(filename=str(out)))
```
